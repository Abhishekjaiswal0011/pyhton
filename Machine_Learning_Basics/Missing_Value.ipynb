```python
# create a sample dataset with missing values
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'A': [1,2,100,4,5], 
    'B': [3,np.nan,7,8,9],
    'C': [np.nan,12,13,14,15],
    'D': [16,17,18,np.nan,20]
}

# Convert the dictionary to a pandas dataframe
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# imputation  of missing value
```

```python
df.isnull().sum() # age is missing at random
```

```python
sns.histplot(df.age,kde = True)
```

```python

# whenever you have normal distribution data and no outliers, impute with mean
df['mean_imputation'] = df['age'].fillna(df['age'].mean())
```

```python
df[['age','mean_imputation']]
```

```python
sns.distplot(df['mean_imputation'], kde = True)
```

```python
# mean imputation works with normally distributed data/no outlier is present
```

```python
# median imputation
df['median_imputation'] = df['age'].fillna(df['age'].median())
df[['age','mean_imputation','median_imputation']]
```
