```python
import pandas as pd 
import numpy as np
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# feature scaling
# standardisation >> mu=0, sigma =1
# normalization(minmax scaling) >> data comes between 0 and 1
# unit vector >> 1
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.distplot(df["total_bill"])
```

```python
mean = np.mean(df['total_bill'])
std = np.std(df["total_bill"])
```

```python
std_data = []
for i in list(df["total_bill"]):
    zscore = (i-mean)/std
    std_data.append(zscore)
```

```python
std_data
```

```python
# feature scaling is optional >> it does't change the distribution of data
```

```python
sns.distplot(df['total_bill'])
```

```python
sns.distplot(std_data)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler
```

```python
scaler.fit(df[['total_bill']]) # it will calculate mu and sigma
```

```python
scaler.transform(df[['total_bill']])
```
