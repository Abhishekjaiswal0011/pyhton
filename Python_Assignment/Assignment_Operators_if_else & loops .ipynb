```python
number = 12321
original = number
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10
    
print(reverse)

if reverse == original:
   print("Palindrome")
else:
   print("Not palindrome")
```
