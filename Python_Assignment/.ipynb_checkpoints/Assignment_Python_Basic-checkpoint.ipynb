```python
total_minutes = 135

# First 15 minutes are free as per mall policy,
# so they are not counted in the bill
free_minutes = 15
billable_minutes = total_minutes - free_minutes

# Mall Charges for every started hour,so round Up
# using - (-a//b) which is ceiling division
hours = -(-billable_minutes // 60)

rate_per_hour = 20
fee = hours * rate_per_hour
print(fee)
```
