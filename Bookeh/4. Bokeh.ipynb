```python
from bokeh.sampledata.iris import flowers
from bokeh.plotting import figure, output_notebook, show

output_notebook()

p = figure(title='test_flower')
p.xaxis.axis_label = "petal_length"
p.yaxis.axis_label = "petal_width"
p.line(flowers['petal_length'], flowers['petal_width'], line_width=2, line_alpha=0.6)
show(p)
```

```python
x = [1, 2, 3, 4]
y = [3, 8, 4, 7]

p = figure(title="line_chart")
p.line(x, y)
show(p)
```

```python
x = [1, 2, 3, 4]
y = [3, 8, 4, 7]

p = figure(title="line_chart")
p.scatter(x, y, size = 20, fill_color = 'red', legend_label = 'my_line')
show(p)
```
